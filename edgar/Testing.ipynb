```python
from edgar import httpclient_ratelimiter

from edgar import httpclient
from pyrate_limiter import Limiter, Duration, Rate

bucket = httpclient_ratelimiter.create_sqlite_rate_limiter("mylimiter4", rate=Rate(10, Duration.SECOND))
httpclient._RATE_LIMITER = Limiter(bucket)

```

```python
from concurrent.futures import ProcessPoolExecutor

companies = ["IBM", "MSFT", "TSLA", "UPS"]

def get_company_filings(company): 
    try:
        httpclient._RATE_LIMITER = httpclient_ratelimiter.create_sqlite_rate_limiter(10, 60000)
        filings = Company(company).get_filings(form="10-Q")
        return filings
    except Exception as e:
        print(e)

f=get_company_filings("IBM")

type(f[0])
```
